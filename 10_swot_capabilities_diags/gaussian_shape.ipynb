```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def gaussian_shape(d, lambda_c):
    norm = np.sqrt(2*np.pi/(np.log(2)*lambda_c**2))
    exp = -2*np.pi**2/np.log(2)*(d/lambda_c)**2
    return norm*np.exp(exp)
```

```python
d = np.arange(-10, 10)
plt.plot(d, gaussian_shape(d, 9))
plt.axvline(-2)
plt.axvline(2)
```
