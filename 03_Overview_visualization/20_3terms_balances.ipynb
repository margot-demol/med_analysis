```python
import numpy as np
import pandas as pd
import xarray as xr

import matplotlib.pyplot as plt

def compute_ds(ds):
    ds['S'] = ds.a+ds.b+ds.c
    ds['X_ab'] = -2*ds.a*ds.b
    ds['X_bc'] = -2*ds.b*ds.c
    ds['X_ac'] = -2*ds.a*ds.c
    ds['B_a'] = - ds.a*ds.b -ds.a*ds.c
    ds['B_b'] = - ds.a*ds.b -ds.b*ds.c
    ds['B_c'] = - ds.a*ds.c -ds.b*ds.c
    ds['E_a'] = ds.a*ds.S
    ds['E_b'] = ds.b*ds.S
    ds['E_c'] = ds.c*ds.S

def compute_ds_var(ds):
    ds['S'] = ds.a+ds.b+ds.c
    ds['X_ab'] = -2*(ds.a*ds.b - ds.a.mean()*ds.b.mean())
    ds['X_bc'] = -2*(ds.b*ds.c - ds.b.mean()*ds.c.mean())
    ds['X_ac'] = -2*(ds.a*ds.c - ds.a.mean()*ds.c.mean())
    ds['B_a'] = - ds.a*ds.b -ds.a*ds.c
    ds['B_b'] = - ds.a*ds.b -ds.b*ds.c
    ds['B_c'] = - ds.a*ds.c -ds.b*ds.c
    ds['E_a'] = ds.a*ds.S
    ds['E_b'] = ds.b*ds.S
    ds['E_c'] = ds.c*ds.S

def plot(ds, dsm):
    fig, axs = plt.subplots(1, 3, figsize = (10, 5), sharey=True)
    i=0
    for v in ['a','b','c']:
        ds[v].plot(ax=axs[i], c=c0[v])
        axs[i].grid()
        i+=1

    fig, ax = plt.subplots(1, 1, figsize = (10, 5))
    i = 0
    for v in ['a','b','c']:
        ax.bar(i, dsm[v], color= c0[v])
        ax.bar(i, dsm['B_'+v], width = 0.2, color= c0[v], edgecolor='k')
        ax.bar(i, dsm['E_'+v], width = 0.2, color= 'grey')
        i+=1
    for v in ['ab', 'ac', 'bc'] :
        ax.bar(i, dsm['X_' + v], color= c0[v])
        i+=1 
    ax.grid()
    return ax
    
c0 = dict(a='b', b='r', c='y', ab='purple', ac = 'green', bc = 'orange')
```

# First case : 2 terms compensate c, no mean

$$a + b + c = 0$$

```python
N = 100000
s0 = (np.random.rand(N)-0.5)*1
s1 = (np.random.rand(N)-0.5)*2
a = s0
b = s1
c = -s0 - s1

ds = xr.Dataset(data_vars=dict(a=('row_number', a),b=('row_number', b),c=('row_number', c),), 
                coords=dict(row_number = ('row_number', np.arange(N))),
               )
compute_ds(ds)
dsm = xr.merge([(ds[['a', 'b', 'c', 'S']]**2).mean(), ds[[v for v in ds if '_' in v]].mean()])
ax = plot(ds, dsm)
ax.set_ylim(0,5)
```

# Second case : 2 terms compensate c, non null mean

- contribution negative de a et b, qui compensent ensemble c mais qui ne sont pas corrélés (X_ab = -2*(a.mean()x b.mean()))

```python
N = 100000
s0 = (np.random.rand(N)-0.1)*1
s1 = (np.random.rand(N)+0.2)*2
a = s0
b = s1
c = -s0 - s1

ds = xr.Dataset(data_vars=dict(a=('row_number', a),b=('row_number', b),c=('row_number', c),), 
                coords=dict(row_number = ('row_number', np.arange(N))),
               )
compute_ds(ds)
dsm = xr.merge([(ds[['a', 'b', 'c', 'S']]**2).mean(), ds[[v for v in ds if '_' in v]].mean()])
ax = plot(ds, dsm)
ax.set_ylim(0,5)
```

# Third case : 3 terms correlated, 0 means

```python
N = 10000
s0 = (np.random.rand(N)-0.5)*2
s1 = (np.random.rand(N)-0.5)*3
s2 = (np.random.rand(N)-0.5)*1
a = s0 + s2
b = s1 - s2
c = -s0 - s1 

ds = xr.Dataset(data_vars=dict(a=('row_number', a),b=('row_number', b),c=('row_number', c),), 
                coords=dict(row_number = ('row_number', np.arange(N))),
               )
compute_ds(ds)
dsm = xr.merge([(ds[['a', 'b', 'c', 'S']]**2).mean(), ds[[v for v in ds if '_' in v]].mean()])
    
ax = plot(ds, dsm)
ax.set_ylim(0,5)
```

# Fourth case : 3 terms correlated, non null means

```python
N = 10000
s0 = (np.random.rand(N)-0.5)*2
s1 = (np.random.rand(N)-0.1)*3
s2 = (np.random.rand(N)-0.5)*1
a = s0 + s2
b = s1 - s2
c = -s0 - s1 

ds = xr.Dataset(data_vars=dict(a=('row_number', a),b=('row_number', b),c=('row_number', c),), 
                coords=dict(row_number = ('row_number', np.arange(N))),
               )
compute_ds(ds)
dsm = xr.merge([(ds[['a', 'b', 'c', 'S']]**2).mean(), ds[[v for v in ds if '_' in v]].mean()])
    
ax = plot(ds, dsm)
ax.set_ylim(0,5)
```

# Corrections
- BUT : how to account for the mean balance ??

```python
ds = xr.Dataset(data_vars=dict(a=('row_number', a),b=('row_number', b),c=('row_number', c),), 
                coords=dict(row_number = ('row_number', np.arange(N))),
               )

bm = ds.b.mean('row_number').values
cm = ds.c.mean('row_number').values
delta = xr.DataArray(data = np.where(abs(bm)>abs(cm), abs(cm), abs(bm)))
    
ds['meanc'] = np.sign(cm)*delta
ds['meanb'] = np.sign(bm)*delta
        
ds['b'] = ds.b-np.sign(bm)*delta
ds['c'] = ds.c-np.sign(cm)*delta

compute_ds(ds)
dsm = xr.merge([(ds[['a', 'b', 'c', 'S']]**2).mean(), ds[[v for v in ds if '_' in v]].mean()])
ax = plot(ds, dsm)
ax.bar(1, (ds.meanb**2).mean(), color='k', bottom = dsm.b)
ax.bar(2, (ds.meanb**2).mean(), color='k', bottom = dsm.c)
ax.bar(5, 2*(ds.meanb**2).mean(), color='k', bottom = dsm.X_bc)
ax.set_ylim(0,5)
```
