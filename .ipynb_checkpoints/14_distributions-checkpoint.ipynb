# PB 

## Too close from the edge of the swath

```python
def plot_exemple(pb, directory):
    for rn in pb.row_number:
        pb_ = pb.sel(row_number=rn)

        dl = 0.2
        bbox = [pb_.longitude.values-dl, pb_.longitude.values+dl, pb_.latitude.values-dl, pb_.latitude.values+dl]
        
        dfc_ = dfc.where((dfc.pass_number==pb_.pass_number.values)&(dfc.cycle_number==pb_.cycle_number.values)&(dfc.drifter_id==pb_.drifter_id.values)).dropna()
        dfwc_ = dfw.loc[rn.values]
        df_ = df.where((df.pass_number==pb_.pass_number.values)&(df.cycle_number==pb_.cycle_number.values)&(df.drifter_id==pb_.drifter_id.values)).dropna()
        dfs = browse_swot_250().reset_index()
        dss = xr.open_dataset(dfs.where((dfs.pass_number==pb_.pass_number.values)&(dfs.cycle_number==pb_.cycle_number.values)).dropna().file.values[0])
        dss = dss.where((dss.latitude>bbox[2]) & (dss.latitude<bbox[3])&(dss.longitude>bbox[0]) & (dss.longitude<bbox[1]))
        
        fig = plt.figure( frameon=False, figsize=(20,6))
        ax = fig.add_subplot(131, projection=ccrs.Orthographic(df.longitude.mean(), df.latitude.mean()))
        
        ax.add_feature(cfeature.LAND,)
        gl = ax.gridlines(draw_labels=True,)
        ax.set_extent(bbox)
        
        #SWOT
        dss['eta'] = dss.cvl_mean_dynamic_topography_cnes_cls_22 + dss.cvl_ocean_tide_fes_2022 + dss.duacs_ssha_karin_2_filtered
        dss.eta.plot(x='longitude', y='latitude', ax=ax, transform=crs,cmap='viridis')
    
        #ERA5
        #t = dss.time.mean()
        #era_ = era.sel(time=dss.time.mean(), method='nearest').sortby('latitude')
        #era_ = era_.sel(longitude=slice(bbox[0], bbox[1]), latitude=slice(bbox[2], bbox[3]))
        #Q = era_.plot.quiver('longitude', 'latitude', 'u10', 'v10', ax=ax, transform =crs, scale=250, color='magenta', clip_on = False, add_guide = False)
        
        #Drifters
        df_.plot.scatter('longitude', 'latitude', transform =crs, ax=ax, s=2 )
        dfc_.plot.scatter('longitude', 'latitude', transform =crs, ax=ax, marker='*', color='r', s=5 )
    
        ax = fig.add_subplot(132)
        df_['sume'] = df_.acce + df_.coriolise + df_.ggrade_eta_swot250gauss1e3 + df_.winde0_era5
        df_ = df_.set_index('datetime')
        df_.acce.plot(ax=ax, label = 'acc', c=c0['acc'], ls='', marker='.')
        df_.coriolise.plot(ax=ax, c=c0['coriolis'], label = 'coriolis', ls='', marker='.')
        df_.ggrade_eta_swot250gauss1e3.plot(ax=ax,c=c0['ggrad'],  label = 'ggrad', ls='', marker='.')
        df_.winde0_era5.plot(ax=ax,c=c0['wind'],  label = 'wind', ls='', marker='.')
        df_.sume.plot(ax=ax, c='k', label = 's', ls='--')
        ax.axvline(dfc_.datetime.values[0], color = 'r',ls=':')
        ax.axvline(dss.time.mean().values, color = 'b',ls=':')
        ax.legend()
        ax.grid()
        ax.set_ylim(-7e-5, 7e-5)
    
    
        ax = fig.add_subplot(133)
        df_['sumn'] = df_.accn + df_.coriolisn + df_.ggradn_eta_swot250gauss1e3 + df_.windn0_era5
        df_.accn.plot(ax=ax, label = 'acc', c=c0['acc'], ls='', marker='.')
        df_.coriolisn.plot(ax=ax, c=c0['coriolis'], label = 'coriolis', ls='', marker='.')
        df_.ggradn_eta_swot250gauss1e3.plot(ax=ax,c=c0['ggrad'],  label = 'ggrad', ls='', marker='.')
        df_.windn0_era5.plot(ax=ax,c=c0['wind'],  label = 'wind', ls='', marker='.')
        df_.sumn.plot(ax=ax, c='k', label = 's', ls='--')
        ax.axvline(dfc_.datetime.values[0], color = 'r',ls=':')
        ax.axvline(dss.time.mean().values, color = 'b',ls=':')
        ax.legend()
        ax.grid()
        ax.set_ylim(-7e-5, 7e-5)
    
        fig.suptitle(f'{pb_.row_number.values}, {pb_.pass_number.values}, {pb_.cycle_number.values}, {pb_.drifter_type.values}, {pb_.drifter_id.values} \n' + f'wind ={np.sqrt(dfwc_.u10**2 + dfwc_.v10**2)}'+ f', idcomb ={DSD.id_comb.values}' )
        fig.savefig(os.path.join(images_dir, directory, f'{rn.values}.png'), dpi=200, bbox_inches='tight')
```

```python
pb = DSS.where((abs(DSS.sume)>abs(DSS.sume).quantile(q))| (abs(DSS.sumn)>abs(DSS.sumn).quantile(q)), drop=True)
plot_exemple(pb, 'exemples_highS')
```

```python
pb = DSS.where((abs(DSS.sume)<abs(DSS.sume).quantile(1-q))| (abs(DSS.sumn)<abs(DSS.sumn).quantile(1-q)), drop=True)
plot_exemple(pb, 'exemples_lowS')
```

```python
pb = DSD.where((abs(DSD.sume)>abs(DSD.sume).quantile(q))| (abs(DSD.sumn)>abs(DSD.sumn).quantile(q)), drop=True)
plot_exemple(pb, 'exemples_highS')
```

```python
pb = DSD.where((abs(DSD.sume)<abs(DSD.sume).quantile(1-q))| (abs(DSD.sumn)<abs(DSD.sumn).quantile(1-q)), drop=True)
plot_exemple(pb, 'exemples_lowS')
```
