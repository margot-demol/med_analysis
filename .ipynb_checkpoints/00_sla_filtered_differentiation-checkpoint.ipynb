# Stencil 9pts kernel filter inspired from products routine
- for to apply : https://docs.scipy.org/doc/scipy/reference/generated/scipy.ndimage.generic_filter.html#scipy.ndimage.generic_filter

```python
def stencil_diff(image, dir_='y'):
    from scipy.ndimage import convolve
    st = -np.array([[1/280, -4/105, 1/5], [4/5, 0, -4/5], [-1/5, 4/105, -1/280]])
    if dir_ == 'y': st = st.T
    return image.dot(st.flatten())
    
image = np.array([[1, 1, 1], [2, 2, 2], [3, 3, 3]])
print(stencil_diff(image.flatten(), dir_='x'), stencil_diff(image.flatten(), dir_='y'))
plt.imshow(np.array([[1, 1, 1], [2, 2, 2], [3, 3, 3]]))
```

```python
def apply_stencil_diff_filter(swot_image, dir_):
    from scipy.ndimage import generic_filter
    swot_image = swot_image.interpolate_na('num_pixels').interpolate_na('num_lines')
    return xr.DataArray(generic_filter(swot_image, stencil_diff, size=[3,3], extra_keywords = {'dir_':dir_}), dims=swot_image.dims)

def stencil_diff_new(image, dir_='y'):
    from scipy.ndimage import convolve
    image = image.interpolate_na('num_pixels').interpolate_na('num_lines')
    st = np.array([[1/280, -4/105, 1/5], [4/5, 0, -4/5], [-1/5, 4/105, -1/280]])
    if dir_ == 'y': st = st.T
    return xr.DataArray(convolve(image, st), dims=image.dims)


def fitting_coeff(image,var): 
    """
    var must be in 'cste', 'dx', 'dy', 'dxx', 'dyy', 'dxy'
    """
    meaning_coeff = {'cste':0, 'dx':1, 'dy':2, 'dxx':3, 'dyy':4, 'dxy':5}
    assert var in meaning_coeff.keys(), "var must be in 'cste', 'dx', 'dy', 'dxx', 'dyy', 'dxy'"
    n=int(np.sqrt(image.shape[0]))
    if isinstance(image, xr.DataArray): image = image.values
    x = np.arange(n)
    y = np.arange(n)
    X, Y = np.meshgrid(x, y, copy=False)
    X, Y = np.meshgrid(x, y, copy=False)

    X = X.flatten()
    Y = Y.flatten()

    A = np.array([X*0+1, X, Y, X**2, Y**2, X*Y**2, X*Y]).T

    coeff, r, rank, s = np.linalg.lstsq(A, image)
    return coeff[meaning_coeff[var]]

def apply_fitting_kernel_filter(swot_image, npts, var):
    from scipy.ndimage import generic_filter
    swot_image = swot_image.interpolate_na('num_pixels').interpolate_na('num_lines')
    return xr.DataArray(generic_filter(swot_image, fitting_coeff, size=[npts, npts], extra_keywords = {'var':var}), dims=swot_image.dims)
```

```python
fig, axs = plt.subplots(2, 4, figsize=(20, 10))
dx, dy = 2000., 2000.

# Stencil a la mano
(apply_stencil_diff_filter(sla, dir_='x')/dx).assign_attrs({'long_name':'Zonal diff'}).plot(ax=axs[0, 0])
(apply_stencil_diff_filter(sla, dir_='y')/dy).assign_attrs({'long_name':'Meridional diff'}).plot(ax=axs[1, 0])
axs[0, 0].set_title('Stencil generic_filter')

#Stencil convolve
(stencil_diff_new(sla, dir_='x')/dx).assign_attrs({'long_name':'Meridional diff'}).plot(ax=axs[0, 1])
(stencil_diff_new(sla, dir_='y')/dy).assign_attrs({'long_name':'Meridional diff'}).plot(ax=axs[1, 1])
axs[0, 1].set_title('Stencil convolve')

# xarray differentiate
(sla.differentiate('num_pixels')/dx).assign_attrs({'long_name':'Meridional diff'}).plot(ax=axs[0, 2])
(sla.differentiate('num_lines')/dy).assign_attrs({'long_name':'Meridional diff'}).plot(ax=axs[1, 2])
axs[0,2].set_title('Xarray differentiate')


#Fitting kernel
(apply_fitting_kernel_filter(sla, 5, 'dx')/dx).assign_attrs({'long_name':'Meridional diff'}).plot(ax=axs[0,3])
(apply_fitting_kernel_filter(sla, 5, 'dy')/dy).assign_attrs({'long_name':'Meridional diff'}).plot(ax=axs[1, 3])
axs[0, 3].set_title('Fitting kernel')

fig.tight_layout()
```

```python
from cstes import zarr_dir
dfs = browse_swot_2km().reset_index()
g=9.81
dx = 2000.
dy = 2000.

for npts in np.arange(3, 21, 2):
    for f in dfs.file.unique() : 
        dss = xr.open_dataset(f)
        dss = add_grid_metrics(dss)
        dsss=xr.Dataset()
        for v in ['duacs_ssha_karin_2_calibrated','duacs_ssha_karin_2_filtered']:
            dsss['dx_'+v] = apply_fitting_kernel_filter(dss[v], npts, 'dx')/dx
            dsss['dy_'+v] = apply_fitting_kernel_filter(dss[v], npts, 'dy')/dy
            dsss['dxx_'+v] = apply_fitting_kernel_filter(dss[v], npts, 'dxx')/dx/dx
            dsss['dyy_'+v] = apply_fitting_kernel_filter(dss[v], npts, 'dyy')/dy/dy
            dsss['dxy_'+v] = apply_fitting_kernel_filter(dss[v], npts, 'dxy')/dx/dy
            dsss = dsss.where(dss.duacs_editing_flag==0)
        print(int(dss.pass_number.mean()), int(dss.cycle_number.mean()))
        dsss.to_netcdf(os.path.join(zarr_dir,'before_coloc', '2km_swot_fitting_kernel', f'{npts}pts_fittingkernel_{int(dss.pass_number.mean())}_{int(dss.cycle_number.mean())}.nc'))
```
