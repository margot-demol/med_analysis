```python
import numpy as np
import pandas as pd
import xarray as xr

import matplotlib.pyplot as plt
```

# Création champ constant sla / vitesse géostrophique

```python
#Choose
Ugaussmax = 0.5
Uf = 0.1
g = 9.81

sx=sy = 20e3
mx=my = 30e3
# Champ de sla = affine + gaussian
# define non-normalized 2D gaussian
def gaus2d(x=0, y=0, mx=0, my=0, sx=1, sy=1):
    return np.exp(-((x - mx)**2. / (2. * sx**2.) + (y - my)**2. / (2. * sy**2.)))

x = np.linspace(-5, 5)
y = np.linspace(-5, 5)
x, y = np.meshgrid(x, y) # get 2D variables instead of 1D
z = gaus2d(x, y)

lx = 100e3
ly = 100e3
xx, yy  = np.meshgrid(np.arange(0, lx, 100), np.arange(0, ly, 100))
aff = f*Uf/g*xx-0.1 # = f Uf /g *x - cste 

aff = xr.DataArray(data = aff, coords=dict( y = (['y'], np.arange(0,ly, 100)), x = (['x'], np.arange(0,lx, 100))), name = 'aff')
A = Ugaussmax/(g/f/sx/np.sqrt(np.exp(1)))
gauss = xr.DataArray(data = A*gaus2d(x=xx, y=yy, mx=mx, my=my, sx=sx, sy=sy), coords=dict( y = (['y'], np.arange(0,ly, 100)), x = (['x'], np.arange(0,lx, 100))), name = 'gauss') 
sla = aff+gauss

ds = xr.merge([sla.rename('sla'), aff, gauss])

g = 9.81
ds['ggdx'] = g*ds.sla.differentiate('x')
ds['ggdy'] = g*ds.sla.differentiate('y')
ds['ggdfx'] = g*ds.aff.differentiate('x')
ds['ggdfy'] = g*ds.aff.differentiate('y')
ds['ggdgaussx'] = g*ds.gauss.differentiate('x')
ds['ggdgaussy'] = g*ds.gauss.differentiate('y')

f = 2 * 2 * np.pi / 86164.1 * np.sin(40 * np.pi / 180)
ds['ug'] = -ds.ggdy/f
ds['vg'] = ds.ggdx/f

ds['ugauss'] = -ds.ggdgaussy/f
ds['vgauss'] = ds.ggdgaussx/f

ds['uf'] = -ds.ggdfy/f
ds['vf'] = ds.ggdfx/f
```

```python
fig, axs = plt.subplots(1, 3, figsize=(20, 5))
ds.sla.plot(ax=axs[0])
ds.ggdx.plot(ax=axs[1])
ds.ggdy.plot(ax=axs[2])
ds.isel(x = slice(0,None,50), y = slice(0,None,50)).plot.quiver( 'x', 'y','ug','vg', color='k', scale = 20, ax=axs[0])

fig.tight_layout()
```

# Trajectoires

```python
t = np.arange(0,5*24*3600, 3600)#seconds
T = 18*3600#seconds
ui = 0.05

def inertial_comp(t, ui, T) :
    phi = np.random.random()*2*np.pi
    ai = T/2/np.pi*ui
    trajx = ai*np.cos(2*np.pi*t/T+phi)
    trajy = ai*np.sin(2*np.pi*t/T+phi)
    traj = xr.Dataset(data_vars = dict(inx = (['t'], trajx),iny = (['t'], trajy)),  coords=dict( t = (['t'],t)))
    traj['inu'] = traj.inx.differentiate('t')
    traj['inv'] = traj.inx.differentiate('t')
    return traj


def geostrophic(traj, x0, y0):
    # init
    ug = ds.ug.interp(x=x0,y=y0).values
    vg = ds.ug.interp(x=x0,y=y0).values
    x, y = x0, y0
    X, Y = [x0], [y0]
    Ug, Vg = [ug], [vg]
    
    # init
    ug = ds.ug.interp(x=x0,y=y0).values
    vg = ds.ug.interp(x=x0,y=y0).values
    dt= traj.t.diff('t').mean().values
    
    for i in range(len(traj.t)):
        traj_= traj.isel(t=i)
       
        x, y = x + traj_.inx.values + ug*dt, y+traj_.iny.values + vg*dt
        
        X.append(x)
        Y.append(y)
        ug = ds.ug.interp(x=x, y=y).values
        vg = ds.vg.interp(x=x, y=y).values

        Ug.append(ug)
        Vg.append(vg)
        
    traj['x'] = (['t'], X[:-1])
    traj['y'] = (['t'], Y[:-1])
    traj['vg'] = (['t'], Vg[:-1])
    traj['ug'] = (['t'], Ug[:-1])

TRAJ = []
#X0 = np.arange(10e3, 100e3,10e3) 
#Y0 =np.ones_like(X0)*45e3
X0 = np.random.rand(100)*lx
Y0 = np.random.rand(100)*ly

for i in range(len(X0)) : 
    traj = inertial_comp(t, ui, T)
    geostrophic(traj, x0=X0[i], y0=Y0[i])
    TRAJ.append(traj)
dst = xr.concat(TRAJ, dim = 'drifter')
    
```

```python
fig, ax = plt.subplots(1, 1, figsize=(7, 7))
ds.sla.plot(ax=ax)
ds.isel(x = slice(0,None,50), y = slice(0,None,50)).plot.quiver( 'x', 'y','ug','vg', color='k', scale = 20, ax=ax)
for d in dst.drifter : 
    dst.sel(drifter=d).plot.scatter(x='x', y='y', ax=ax, marker='.')
fig.tight_layout()
```

```python
D =[]
for d in dst.drifter:
    D.append(ds.interp(x=('t',dst.sel(drifter=d).x.values), y=('t',dst.sel(drifter=d).y.values)).assign_coords(t=('t', dst.t.values)).reset_coords(['x', 'y']))
```

```python
dsat = xr.concat(D, dim ='drifter')
dsat['t_day'] = dsat.t/3600/24
dsat = dsat.set_coords('t_day')
```
